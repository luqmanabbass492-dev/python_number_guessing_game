```python
# number guessing game
import random
random.randint(1,100)

jackpot = random.randint(1, 100)

guess = int(input("guess the number: "))
counter = 1
while guess != jackpot:
    if guess < jackpot:
        print('guess higher')
    else:
        print('guess lower')
    
    
    guess = int(input('go and guess the number: '))
    counter+=1

print('this is the right answer', jackpot)
print("you took",counter,"attempts")
```
